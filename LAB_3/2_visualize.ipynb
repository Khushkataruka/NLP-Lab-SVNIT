```python
import pandas as pd
df=pd.read_csv('word_freq.csv')
df.head()
```

```python
sort_df=df.sort_values(by='freq',ascending=False)
sort_df.head()
display(len(sort_df))
```

```python
import matplotlib.pyplot as plt
from matplotlib import font_manager as fm

# Path to the Noto Sans Devanagari font on Windows
hindi_font_path = "Noto_Sans_Devanagari/NotoSansDevanagari-VariableFont_wdth,wght.ttf"
hindi_font = fm.FontProperties(fname=hindi_font_path, size=12)

plt.figure(figsize=(15, 6))
plt.bar(sort_df["word"][:100].astype(str), sort_df["freq"][:100])
plt.xticks(rotation=90, fontsize=10, fontproperties=hindi_font)
plt.xlabel("शब्द", fontproperties=hindi_font)
plt.ylabel("आवृत्ति", fontproperties=hindi_font)
plt.yscale("log")
plt.title("शीर्ष 100 शब्द आवृत्तियाँ", fontproperties=hindi_font)
plt.tight_layout()
plt.show()
```

```python
# remove stop words
threshhold=5*1e6
sort_df.drop(sort_df[sort_df["freq"] > threshhold].index, inplace=True)
len(sort_df)
```

```python


plt.figure(figsize=(15, 6))
plt.bar(sort_df["word"][:100].astype(str), sort_df["freq"][:100])
plt.xticks(rotation=90, fontsize=10, fontproperties=hindi_font)
plt.xlabel("शब्द", fontproperties=hindi_font)
plt.ylabel("आवृत्ति", fontproperties=hindi_font)
plt.yscale("log")
plt.title("शीर्ष 100 शब्द आवृत्तियाँ", fontproperties=hindi_font)
plt.tight_layout()
plt.show()
```
